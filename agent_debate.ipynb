```python
rounds = 1
```

```python
import time
for i in range (rounds):
    prompt_agent1 = f"""
You are participating in a debate.
You are Agent 1.

Your personality:
(Agent 1 personality text)

Rules:
(Agent 1 rules)

Debate topic:
does a country need immigrants

Conversation so far:
{memory}

Now give your argument.
"""
    responce_agent1_inloop = model.generate_content(prompt_agent1)
    print(f"agent1:{responce_agent1_inloop.text}")
    memory = memory + "\nagent 1: " + responce_agent1_inloop.text
    time.sleep(6)

    # MAX_char= 1500
    # if len(memory)> MAX_char:
    #     memory = memory[-MAX_char]

    print("    ")
    print("   ")


    prompt_agent2 = f"""
You are participating in a debate.
You are Agent 2.

Your personality:
(Agent 2 personality text)

Rules:
(Agent 2 rules)

Debate topic:
does a country need immigrants

Conversation so far:
{memory}

Now give your counter-argument.
""" 
    responce_agent2_inloop = model.generate_content(prompt_agent2)
    print(f"agent2:{responce_agent2_inloop.text}")
    memory = memory +"\nagent 2 : " + responce_agent2_inloop.text
    time.sleep(6)

    # if len(memory)> MAX_char:
    #     memory = memory[-MAX_char]
```

```python
prompt_judge = f"""You are a neutral debate judge.

Your task is to evaluate a debate objectively and logically.

Debate topic:
does a country need immigrants

Debate transcript:
{memory}

Evaluation criteria:
1. Logical strength of arguments
2. Use of reasoning and evidence
3. Ability to directly rebut the opponent
4. Consistency of position
5. Overall persuasiveness

Instructions:
- Do NOT take sides emotionally.
- Ignore confidence or aggression; focus on reasoning.
- Identify weaknesses in BOTH sides.
- Declare ONE clear winner.

Output format:
1. Brief summary of Agent 1s position
2. Brief summary of Agent 2s position
3. Key strengths and weaknesses of each
4. Final verdict: Agent 1 or Agent 2 (with reason)
"""
```
